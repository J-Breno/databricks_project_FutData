#### Datas

```python
from pyspark.sql.types import StructType, StructField, StringType, DateType
```

```python
esquema = StructType([
    StructField('Nome', StringType(), True),
    StructField('Função', StringType(), True),
    StructField('Data Início', StringType(), True)
])

dados = [
    ("Alberto", "Engenheiro de dados", "27/01/2023"),
    ("Ana", "Cientista de dados", "15/02/2023"),
    ("Beto", "Analista de dados", "01/03/2023")
]

df = spark.createDataFrame(dados, schema=esquema)
```

```python
display(df)
```

```python
from pyspark.sql.functions import to_date
```

```python
df = df.withColumn('Data Início (data)', to_date(df['Data Início'], 'dd/MM/yyyy'))
```

```python
display(df)
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import to_timestamp
```

```python
df = df.withColumn('Data Início (timestamp)', to_timestamp(df['Data Início'], 'dd/MM/yyyy'))
```

```python
display(df)
```

```python
df.printSchema()
```

```python
df = df.withColumn('Data Início (timestamp -> data)', to_date(df['Data Início (timestamp)'], 'dd/MM/yyyy'))
```

```python
display(df)
```

```python
from pyspark.sql.functions import date_format
```

```python
df = df.withColumn('Data Início (string)', date_format(df['Data Início (data)'], 'dd.MM.yyyy'))
```

```python
display(df)
```

```python
esquema = StructType([
    StructField('Nome', StringType(), True),
    StructField('Função', StringType(), True),
    StructField('Data Início', StringType(), True),
    StructField('Data Fim', StringType(), True),
])

dados = [
    ('Alberto', 'Engenheiro de dados', '27/01/2020', '27/01/2021'),
    ('Ana', 'Analista de dados', '12/03/2020', '27/01/2023'),
    ('Beto', 'Cientista de dados', '27/01/2020','17/05/2022')
]

df = spark.createDataFrame(dados, schema=esquema)
```

```python
df = df \
    .withColumn('Data Início', to_date(df['Data Início'], 'dd/MM/yyyy')) \
    .withColumn('Data Fim', to_date(df['Data Fim'], 'dd/MM/yyyy'))
```

```python
display(df)
```

```python
from pyspark.sql.functions import datediff, col
```

```python
df = df.withColumn('Diff Datas', datediff(col('Data Fim'), col("Data Início")))
```

```python
display(df)
```

```python
from pyspark.sql.functions import date_add, date_sub
```

```python
df = df \
    .withColumn('Data Início + 7', date_add(df['Data Início'], 7)) \
    .withColumn('Data Fim - 3', date_sub(df['Data Fim'], 3))
```

```python
display(df)
```

```python
from pyspark.sql.functions import year, month, dayofmonth, hour, minute, second
```

```python
esquema = StructType([
    StructField('Nome', StringType(), True),
    StructField('Função', StringType(), True),
    StructField('Data Início', StringType(), True)
])

dados = [
    ("Alberto", "Engenheiro de dados", "27/01/2023 10:00:00"),
    ("Ana", "Cientista de dados", "15/02/2023 15:41:10"),
    ("Beto", "Analista de dados", "01/03/2023 23:30:30")
]

df = spark.createDataFrame(dados, schema=esquema)
```

```python
df = df.withColumn('Data Início (timestamp)', to_timestamp(df['Data Início'], 'dd/MM/yyyy HH:mm:ss'))
```

```python
df = df \
    .withColumn('Ano', year(df['Data Início (timestamp)'])) \
    .withColumn('Mês', month(df['Data Início (timestamp)'])) \
    .withColumn('Dia', dayofmonth(df['Data Início (timestamp)'])) \
    .withColumn('Hora', hour(df['Data Início (timestamp)'])) \
    .withColumn('Minuto', minute(df['Data Início (timestamp)'])) \
    .withColumn('Segundo', second(df['Data Início (timestamp)']))
```

```python
display(df)
```
