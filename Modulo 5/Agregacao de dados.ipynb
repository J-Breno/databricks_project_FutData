```python
partidas_df = spark.read.parquet(f'{caminho_gold}fato_partidas')
```

```python
display(partidas_df)
```

```python
from pyspark.sql.functions import count, countDistinct, max, min, sum, avg
```

```python
partidas_df.select(count("*")).display()
```

```python
partidas_df.select(countDistinct("ano_torneio")).display()
```

```python
partidas_df.select('ano_torneio').distinct().display()
```

```python
resultados_df = (
    partidas_df.filter('nr_partida = "Regular season - 1"')
    .select(sum(partidas_df.qt_gols_casa), sum(partidas_df.qt_gols_fora))
    .withColumnRenamed("sum(qt_gols_casa)", "total_gols_casa")
    .withColumnRenamed("sum(qt_gols_fora)", "total_gols_fora")
)
```

```python
resultados_df.display()
```

```python
partidas_df.groupBy("ano_torneio").agg(
    sum("qt_gols_casa").alias('total_gols_casa'), sum("qt_gols_fora").alias('total_gols_fora'), avg("qt_gols_casa").alias('media_gols_casa')
).display()
```
