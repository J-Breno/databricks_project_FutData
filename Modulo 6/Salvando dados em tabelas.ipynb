```python
jogadores_df = spark.read.parquet(f'{caminho_silver}jogadores')
```

```python
jogadores_df.write.parquet(f'{caminho_gold}dim_jogadores', mode='overwrite')
```

```python
jogadores_df.createOrReplaceTempView('view_jogadores')
```

```python
%sql

SELECT * FROM view_jogadores
```

```python
%sql

select
    id_jogador,
    nm_jogador,
    CAST(DATEDIFF(YEAR, CAST(dados_nascimento.date AS DATE), CURRENT_DATE()) AS INT) AS idade,
    concat(coalesce(dados_nascimento.place, 'Não definido'), ', ', nm_nacionalidade) AS local_nascimento,
    nm_nacionalidade as pais_nascimento,
    altura,
    peso,
    link_foto
from
    view_jogadores
```

```python
%sql

CREATE OR REPLACE TABLE futdata.dim_jogadores
AS
select
    id_jogador,
    nm_jogador,
    CAST(DATEDIFF(YEAR, CAST(dados_nascimento.date AS DATE), CURRENT_DATE()) AS INT) AS idade,
    concat(coalesce(dados_nascimento.place, 'Não definido'), ', ', nm_nacionalidade) AS local_nascimento,
    nm_nacionalidade as pais_nascimento,
    altura,
    peso,
    link_foto
from
    view_jogadores
```
